```python
import requests
import sys

#both packages are native to anaconda
#to install new packages use !pip install {package name} - pip is an install package manager
```

```python
brewery_type = {} #dictionary variable type
state = input("Enter a state name: ") 
#input is a function that prompts a user for some kind of input
#extra controls can be added if desired
city = input("Enter a city name: ")
url = "https://api.openbrewerydb.org/breweries?format=json&by_state=" + state + "&by_city=" + city
#any url with a ? indicates it's sending some kind of data in a key-value pair (dictionary)
response = requests.get(url)
#requests package allows us to access the data from the API
brew = response.json()
#returns the response as a json object instead of the json-formatted text file
```

```python
#print(type(brew))
#type function shows the brew variable type
#print(type(brew[1]))
#type of the index within the variable brew

#function counts the amounts of different brewery_type values as they exist within the JSON dictionary
for i in range(len(brew)):
    #i is a declaration of the integer variable for the index number - i is short for integer
    #range is created based on the length of brew - how many elements are in the brew variable
    if brew[i]["brewery_type"] in brewery_type:
        brewery_type[brew[i]["brewery_type"]] += 1
        #adds 1 to the value if the same brewery type exists
    else:
        brewery_type.update({brew[i]["brewery_type"]: 1})
        #puts 1 as the value for the new brewery type

#the first run will trigger the else statement as the first element has nothing in it
```

```python
print("\nTypes of breweries in " + city + ", " + state + ":\n" + str(brewery_type) + "\n")
#prints out the types of breweries in the user's selected state, 
#with the dictionary variable printed as a string on a separate line
```
