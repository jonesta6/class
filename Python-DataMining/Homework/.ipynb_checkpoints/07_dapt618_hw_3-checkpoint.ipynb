# 1. Plot histograms for all variables in this data set. Add a vertical line to show the average value of each variable. Add a chart title for each chart. [5 points]


Loading in the data and taking a first look at it

```python
import pandas as pd

wine_df = pd.read_csv("https://raw.githubusercontent.com/jonesta6/class/refs/heads/tom/class/datamining/Python-DataMining/Data%20Files/winequality.csv", encoding='ISO-8859-2')

wine_df.info()
```

```python
wine_df['quality'].value_counts()
```

Performing some cleansing operations

```python
wine_df = wine_df.drop(wine_df.columns[0], axis=1)
wine_df.columns = wine_df.columns.str.replace(' ', "", regex=False)
wine_df.head()
```

Build histograms

```python
# Import necessary packages
import matplotlib.pyplot as plt
import seaborn as sns

# Declare variables
wine_var_df = wine_df.iloc[:, :-2]
num_cols = wine_var_df.columns
BINS = 100

fig, axes = plt.subplots(nrows=6, ncols=2, figsize=(12, 9))
plt.tight_layout(pad = 2.0)

# Iterate through axes 
for i, ax in enumerate(axes.flat[:len(num_cols)]):
    col = num_cols[i]
    wine_var_df[col].hist(bins=BINS, ax=ax, grid = False)
    mean_value = wine_var_df[col].mean()
    ax.axvline(mean_value, color='black')
    ax.set_xlabel(col, weight='bold', size=10)

# Remove blank plot
for ax in axes.flat[len(num_cols):]:
    fig.delaxes(ax)

plt.show()
```

# 2. Identify two features that are most correlated with each other. Visualize their relationship using an appropriate chart type. [3 points]

Develop a correlation matrix and view the appropriate heatmap to find highly correlated features

```python
corr_matrix = wine_df.corr(numeric_only=True)

cmap = sns.diverging_palette(10, 220, n=20)
plt.figure(figsize=(11,9))
sns.heatmap(corr_matrix, cmap = cmap);
```

The most correlated features are free sulphur dioxide and total sulphur dioxide. Creating a scatterplot to show their relationship.

```python
plt.figure(figsize=[12,9])

ax = sns.regplot(x='freesulfurdioxide', y='totalsulfurdioxide', data=wine_df)

plt.xlabel('Free Sulfur Dioxide', fontsize=14, weight='bold')
plt.ylabel('Total Sulfur Dioxide', fontsize=14, weight='bold');
```

# 3. Identify the feature that is most correlated with wine quality and plot their relationship using an appropriate chart type. Use different colors to differentiate between red and white wines. [4 points]

Based on the heatmap of correlations above, alcohol is the most correlated feature to quality. Building a scatterplot of the two.

```python
plt.figure(figsize=[12,9])

ax = sns.scatterplot(x='quality', y='alcohol', data=wine_df, color='tomato', s=100, hue='winetype')

plt.xlabel('Quality Score', fontsize=14, weight='bold')
plt.ylabel('Alcohol Content', fontsize=14, weight='bold')
plt.legend(loc='right', fontsize=12);
```

# 4. Bin Fixed Acidity into low, medium, and high categories. Show the distribution of Wine Quality for each acidity category. [3 points]

Get summary statistics for the fixed acidity column to determine how to bin each category. 

```python
wine_df['fixedacidity'].describe()
```

Easiest method to bin categories seems to be based on IQR as middle. Create bins.

```python
Q1 = wine_df['fixedacidity'].quantile(0.25)
Q3 = wine_df['fixedacidity'].quantile(0.75)

wine_df['fixedacidity_bin'] = pd.cut(
    wine_df['fixedacidity'],
    bins=[-float('inf'), Q1, Q3, float('inf')],
    labels=['low', 'medium', 'high']
)

wine_df.fixedacidity_bin.value_counts()
```

Plot out the distributions

```python
sns.displot(wine_df, x='quality', hue='fixedacidity_bin', element='step', bins=50);

```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Define the order of fixedacidity_bin levels
bin_order = ['low', 'medium', 'high']

# Define a unique solid color for each bin level
plot_colors = {
    'low': 'skyblue',
    'medium': 'seagreen',
    'high': 'salmon'
}

# Get sorted quality levels 
qualities = sorted(wine_df['quality'].unique())

# Create a subplot for each fixedacidity_bin level
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)
plt.tight_layout(pad=2.0)

for ax, level in zip(axes, bin_order):
    subset = wine_df[wine_df['fixedacidity_bin'] == level]
    
    # Create a palette dictionary mapping every quality to the same solid color for this plot.
    palette_dict = {q: plot_colors[level] for q in qualities}
    
    # Use countplot with hue set to quality and provide the palette dictionary.
    sns.countplot(
        x='quality',
        hue='quality',  # Assigning hue to avoid the deprecation warning
        data=subset,
        order=qualities,
        palette=palette_dict,
        dodge=False,    # Keep bars together (not side-by-side)
        ax=ax
    )
    
    # Remove the legend since hue duplicates x.
    if ax.get_legend():
        ax.get_legend().remove()
    
    ax.set_title(f'Fixed Acidity: {level}', fontsize=12, weight='bold', pad=10)
    ax.set_xlabel('Quality')
    ax.set_ylabel('Count')

plt.tight_layout()
plt.show()

```

# 5. Create a dataframe (see example below) that contains the overall average for each variable as well as its average value for Red and White wines. Divide both averages by the overall average to create an index for each variable. Multiply the index columns by 100. Which two variables have the highest index? [5 points]

|                 | Red      | White    | Overall  | White_Index | Red_Index  |
|-----------------|----------|----------|----------|-------------|------------|
|fixed acidity	  | 8.319637 | 6.854788 | 7.215307 | 95.003409   | 115.305381 |
|volatile acidity |	0.527821 | 0.278241	| 0.339666 | 81.916094   | 155.3939   |
|.................|..........|..........|..........|.............|............|

```python
# Compute overall means for numeric features
overall_means = wine_df.mean(numeric_only=True)

overall_means

# Compute means for each wine type (assuming a column "wine_type" exists)
wine_type_means = wine_df.groupby('winetype').mean(numeric_only=True)

# Convert overall_means into a DataFrame with features as rows
overall_df = overall_means.to_frame(name='Overall')

# Transpose wine_type_means so that features are rows and wine types are columns
wine_type_df = wine_type_means.T

# Combine the overall means with the wine type means
mean_df = overall_df.join(wine_type_df)

# Add columns for the indexes

mean_df['Red_Index'] = (mean_df['Red'] / mean_df['Overall']) * 100
mean_df['White_Index'] = (mean_df['Red'] / mean_df['Overall']) * 100

mean_df
```
